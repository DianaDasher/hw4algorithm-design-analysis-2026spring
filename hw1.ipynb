## 问题 2

观察下方的分治算法（divide-and-conquer algorithm）的伪代码, 回答下面问题

```latex
DoSomething(A,p,r)
-----
n := r-p+1
if n=2 and A[p]>A[r] then
    swap A[p] and A[r]
else if n >= 3 then
    m = ceil(2n/3)
    DoSomething(A,p,p+m-1)
    DoSomething(A,r-m+1,r)
    DoSomething(A,p,p+m-1)  
    
---
first call: DoSomething(A,1,n)
```

note：$ceil(2n/3)=\left\lceil {2n/3} \right\rceil$；$:=$ 表示赋值, 等价于 $\to$；A是一个包含n的整数元素的数组, 

1. 写出该算法时间复杂度的递归公式, 并求解其对应的渐进表示
2. 描述一下该算法的功能, 并判断是否是最高效的解决方案
3. \* 使用python编程实现上述算法或其对应的更高效版本的算法

idea:\
解：
1.  基本情况：\
        &emsp; 当n=1时，无任何操作: $T(1)=O(1)$ \
        &emsp; 当n=2时，执行一次比较和可能交换操作: $T(2)=O(1)$ \
    递归情况（n>=3）：\
        &emsp; 算法进行了3次递归调用。每次递归调用的子数组大小为 $ m = \lceil 2n/3 \rceil $（约为原始大小的 $2/3$）, 除了递归调用外，计算$n$、$m$以及比较和交换元素所需的时间都是常数时间$O(1)$\
        &emsp; 因此，递归公式可写为：$T(n) = 3T(2n/3) + O(1)$ \
    则整个算法的时间复杂度递归公式为：
    $$T(n) = 3T(2n/3) + O(1) ,  T(1) = 1$$
    \
    求解：符合主定理的标准形式：$T(n)=aT(n/b)+f(n)$, 其中$a=3, b=2, f=O(1)$\
        &emsp; &emsp; 比较：$\log_b a = \log_{\frac{3}{2}} 3 = \frac{\ln 3}{\ln 3 - \ln 2} \approx 2.709$ \
        &emsp; &emsp; $f(n) = 1 = O(n^{\log_{\frac{3}{2}} 3 - \varepsilon})$（取 $\varepsilon \approx 2.709 > 0$），满足主定理情况1;\
        &emsp; &emsp; 因此，渐进表示为：$$T(n) = O\left(n^{\log_{\frac{3}{2}} 3}\right)$$
2. 算法的功能：\
    该算法是一种基于分治思想的排序算法，功能是将数组 $A$ 中的元素按非降序排序。\
    核心工作原理：如果数组末尾的元素小于开头的元素，则交换它们。如果数组包含3个或更多元素，则递归地对前$2/3$排序，然后对后$2/3$排序，最后再对前$2/3$进行一次排序，以确保所有元素都在正确的位置。\
    是否为最高效的解决方案：\
    不是，它的效率极低。 该算法的时间复杂度约为 $O\left(n^{\log_{\frac{3}{2}} 3}\right) \approx O(n^{2.71})$。这比常见的简单排序算法（如冒泡排序、插入排序，复杂度为 $O(n^2)$）还要慢得多。在实际应用中，最高效的基于比较的排序算法。
3. 更高效版本算法：归并排序：将数组均匀分成两半，分别排序后再合并，复杂度为 $O(nlogn)$


````python
# add your code here
# 上述排序算法：
def do_something(A, p, r):
    n = r - p + 1
    if n == 2 and A[p] > A[r]:
        A[p], A[r] = A[r], A[p]
    elif n >= 3:
        m = (2 * n + 2) // 3  # 等价于ceil(2n/3)的整数运算实现
        do_something(A, p, p + m - 1)
        do_something(A, r - m + 1, r)
        do_something(A, p, p + m - 1)

# 更高效的算法：归并排序
def merge_sort(A):
    if len(A) <= 1:
        return A
    mid = len(A) // 2
    left = merge_sort(A[:mid])
    right = merge_sort(A[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# 测试程序
if __name__ == "__main__":
    test = [3, 9, 4, 2, 5, 10, 6, 7, 52, 1, 8, 23, 12, 21, 66]
    print("排序前:", test)
    A = test.copy()
    do_something(test, 0, len(test)-1)
    print("题目算法排序后:", test)
    A = merge_sort(A)
    print("归并排序:", A)
# your algorithm time complexity is:
# 归并排序：将数组均匀分成两半，分别排序后再合并，复杂度为 O(nlogn)
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:
1. 算法设计：枚举法\
&emsp;profits数组中的元素代表的是每天的收益变化量，在第 $i$ 天买入，在第 $j$ 天（$j \ge i$）卖出，所获得的总利润是从第 $i$ 天到第 $j$ 天的收益数组之和：$\sum_{k=i}^{j} profits[k]$，问题转化为寻找数组中的最大子数组和。\
&emsp; 算法设计：使用两个嵌套循环：外层循环遍历所有可能的买入日期 $i$，内层循环遍历在此之后的每一天 $j$ 作为卖出日期。计算每一对 $(i, j)$ 之间的利润总和，并记录下最大值及对应的日期。\
&emsp; 算法时间复杂度分析：需要两层嵌套循环，因此时间复杂度为 $O(n^2)$

````python
# add your code here
def max_profit(profits):
    l = len(profits)
    if l == 0:
        return 0, -1, -1

    max_profit = float('-inf')
    buy_day = 0
    sell_day = 0

    for i in range(l):
        sum = 0
        for j in range(i, l):
            sum += profits[j]
            if sum > max_profit:
                max_profit = sum
                buy_day = i
                sell_day = j
    return max_profit, buy_day + 1, sell_day + 1

# 测试程序
if __name__ == '__main__':
    profits = [3,2,1,-7,5,2,-1,3,-1]
    profit, buy_day, sell_day = max_profit(profits)
    print(f"profits={profits}")
    print(f"第{buy_day}天买入，第{sell_day}天卖出，收益最大：{profit}")
# your algorithm time complexity is:  O(n^2)
````

idea:\
2. 设计时间复杂度为 $O(n)$的算法：\
&emsp; 在遍历数组的过程中，维护一个当前子数组和 now_sum：\
&emsp; 对于遍历到的每一天 profits[i]，选择：是将其加入之前的连续子数组中，或者是从今天开始重新计算。\
&emsp; 如果 now_sum + profits[i] < profits[i]，说明之前的累加减少了今天的收益，则从今天重新开始算起；否则就继续累加。\
&emsp; 在每一步累加后都更新一次最大利润 (max_profit)，并记录下对应的起始和结束天数。\
算法时间复杂度：只需要遍历一次数组，每次遍历中只进行常数次的比较和赋值操作，时间复杂度为 $O(n)$。


````python
def max_profit2(profits):
    l = len(profits)
    if l == 0:
        return 0, -1, -1

    max_profit = float('-inf')
    now_sum = 0
    
    # 记录日期的索引
    buy_index = 0
    sell_index = 0
    temp_buy_index = 0  # 当前子数组的起点

    for i in range(l):
        if profits[i] > now_sum + profits[i]:
            now_sum = profits[i]
            temp_buy_index= i  # 更新潜在的新起点
        else:
            now_sum += profits[i]
            
        # 如果当前累计收益超过了历史最高记录，更新全局最高记录和对应的日期
        if now_sum > max_profit:
            max_profit = now_sum
            buy_index = temp_buy_index
            sell_index = i

    return max_profit, buy_index + 1, sell_index + 1

# 测试程序
if __name__ == '__main__':
    profits = [3,2,1,-7,5,2,-1,3,-1]
    profit, buy_day, sell_day = max_profit2(profits)
    print(f"profits={profits}")
    print(f"第{buy_day}天买入，第{sell_day}天卖出，收益最大：{profit}")
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:
1. 算法设计思路：二分查找\
&emsp; 使用二分查找在$O(logn)$的时间复杂度下找到一个局部最小：通过检查中间元素与其相邻元素的大小关系，来决定下一步搜索左半区间还是右半区间：\
&emsp; 取中点mid，检查$A[mid]$ 是否满足局部最小条件：$A[mid] < A[mid-1]$ 且 $A[mid] < A[mid+1]$ 。如果是，直接返回该值。\
&emsp; 若$A[mid] > A[mid+1]$：则当前在mid处呈下降趋势。因为数组末尾最终会呈现上升趋势（$A[n-2] < A[n-1]$），所以从mid到数组末尾之间，必然存在至少一个局部最小。因此将搜索区间缩小到右半部分；\
&emsp; 若$A[mid] > A[mid-1]$：则当前在mid处呈上升趋势。因为数组开头是下降趋势（$A[0] > A[1]$），所以从数组开头到mid之间，必然存在至少一个局部最小。因此将搜索区间缩小到左半部分。\
算法时间复杂度：$O(\log n)$

````python
# add your code here
# 使用二分查找找到一个局部最小值
def findOneByBinarySearch(A):
    l = len(A)
    # 两端端点不可能是局部最小，所以搜索范围是1到n-2
    left = 1
    right = l - 2
    
    while left <= right:
        mid = (left + right) // 2
        # 检查mid是否为局部最小
        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]
        # 若呈现下降趋势，局部最小值一定在右侧
        elif A[mid] > A[mid + 1]:
            left = mid + 1
        # 若呈现上升趋势，局部最小值一定在左侧
        else:
            right = mid - 1
    return None

# 测试程序
if __name__ == "__main__":
    A=[9, 3, 7, 2, 1, 4, 5 ]
    a= findOneByBinarySearch(A)
    print(f"找到其中一个局部最小值: {a}")
# your algorithm time complexity is: O(logn)
````

idea：\
2. 找出所有局部最小值算法设计：直接对数组进行一次的遍历，检查每个元素是否比它左右两边的邻居都小\
&emsp; 算法时间复杂度： $O(n)$ 

````python
# 线性遍历查找所有局部最小值
def find_all(A):
    n = len(A)
    min= []
    # 遍历除去首尾的中间元素
    for i in range(1, n - 1):
        if A[i] < A[i - 1] and A[i] < A[i + 1]:
            min.append(A[i])
    return min

# 测试程序
if __name__ == "__main__":
    A=[9, 3, 7, 2, 1, 4, 5 ]
    all = find_all(A)
    print(f"所有局部最小元素为: {all}")
````

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:
1. 算法设计：使用快速排序实现：
&emsp; 选择基准：从数组中挑出一个元素作为基准; \
&emsp; 划分：遍历数组，将比基准小的元素放到左边，比基准大的元素放到右边，与基准相等的放中间; \
&emsp; 递归：递归地对左边和右边的子数组进行快速排序; \
&emsp; 拼接：将排序好的左子数组、中间相等的元素和排序好的右子数组拼接起来，得到完整的有序数组;\
&emsp; 截取：最后直接截取有序数组的前$k$个元素。

&emsp; 平均时间复杂度：$O(n \log n)$

````python
# add your code here
#输出最小K个元素
def quick_sort(arr):
    # 数组为空或只有一个元素时，本身就是有序的
    if len(arr) <= 1:
        return arr
        
    # 选择基准：选择中间元素
    p = arr[len(arr) // 2]
    
    # 划分
    left = [x for x in arr if x < p]    # 小于基准的放左边
    mid = [x for x in arr if x == p]    # 等于基准的放中间
    right = [x for x in arr if x > p]   # 大于基准的放右边
    
    # 递归排序左半部分和右半部分并拼接
    return quick_sort(left) + mid + quick_sort(right)

def find_top_k(arr, k):
    if k <= 0 or not arr:
        return []
    after_arr = quick_sort(arr)
    # 截取前 k 个元素
    return after_arr[:k]


# 测试程序
if __name__ == '__main__':
    arr = [5,4,3,2,6,1,88,33,22,107]
    k = 3
    res = find_top_k(arr, k)

    print(f"原始数组: {arr}")
    print(f"快速排序后前 {k} 个最小数据为): {res}")
# your algorithm time complexity is:  O(nlogn)
````

idea：\
2. 算法分析：使用往空堆中不断插入元素的方法建堆:\
   &emsp; 建堆： 初始化一个空堆，将数组中的 $n$ 个元素逐个插入堆中。每次插入的时间复杂度为 $O(\log i)$（ $i$ 是当前堆的大小）;\
   &emsp; 提取： 连续调用 $k$ 次 Extract-min（弹出堆顶最小元素）。每次弹出的时间复杂度为 $O(\log n)$\
   时间复杂度：\
   &emsp; 逐个插入建堆： $\sum_{i=1}^{n} O(\log i) = O(n \log n)$\
   &emsp; 提取 $k$ 次： $k \times O(\log n) = O(k \log n)$\
   总时间复杂度： $$O(n \log n + k \log n) = O(n \log n)$$

````python
import heapq

def findkByMinHeap(arr, k):
    if k <= 0 or not arr:
        return []
        
    min_heap = []
    
    # 往空堆中不断插入元素建堆
    for num in arr:
        heapq.heappush(min_heap, num)
        
    # 调用k次 Extract-min
    res = []
    for _ in range(min(k, len(arr))):
        res.append(heapq.heappop(min_heap))
        
    return res

# 测试程序
if __name__ == '__main__':
    arr = [5,4,3,2,6,1,88,33,22,107]
    k = 3
    res = findkByMinHeap(arr, k)

    print(f"原始数组: {arr}")
    print(f"最小堆法后 {k} 个最小数据为: {res}")
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

算法设计思路：使用一个栈来模拟星球的存续状态，栈底到栈顶对应着从左到右依次存活下来的星球。\
&emsp;从左到右遍历给定的星球数组$A$，对于每一个遇到的星球p：\
&emsp;发生碰撞：当前遍历到的星球p是自右向左运动的（p<0），而栈顶的星球是自左向右运动的（stack[-1] > 0）；\
&emsp;碰撞处理：比较质量\
&emsp;&emsp;若栈顶星球质量<当前星球质量：栈顶星球被摧毁，将其弹出栈；p继续与栈顶星球碰撞；\
&emsp;&emsp;若栈顶星球质量==当前星球质量：向左走的消失，向右走的保留，因此当前星球p被摧毁，判定结束；\
&emsp;&emsp;若栈顶星球质量>当前星球质量：当前星球p被摧毁，判定结束。\
&emsp; 安全入栈：\
&emsp;&emsp;若当前星球p经过了一系列的碰撞依然存活，则将p压入栈中。\

事件复杂度：$O(n)$\
&emsp;虽然在 for 循环内部嵌套了一个 while 循环，但每个星球最多被压入栈一次，也最多被弹出栈一次。因此，所有星球发生的出入栈操作总次数不会超过 $2n$。时间复杂度为线性的 $O(n)$。

idea:

````python
# add your code here
#碰撞算法设计
def collision(A):
    stack = []
    
    for p in A:
        flag = True  #判断p是否存活
        
        # 发生碰撞
        while flag and p < 0 and stack and stack[-1] > 0:
            m_top = abs(stack[-1])
            m_p = abs(p)
            
            if m_top < m_p:
                # 栈顶被摧毁，弹出栈顶，当前星球p 续和下一个栈顶比较
                stack.pop()
            elif m_top == m_p:
                # 质量相同：自右向左p消失，自左向右栈顶元素保留
                flag = False
                break
            else:
                # 栈顶质量大，当前星球p被摧毁
                flag = False
                break
                
        # p依然存活，则入栈
        if flag:
            stack.append(p)
            
    return stack

# 测试程序
if __name__ == '__main__':
    A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
    print(f"输入: {A1}\n输出: {collision(A1)}\n")
    
    A2 = [23, -8, 9, -3, -7, 9, -23, 22]
    print(f"输入: {A2}\n输出: {collision(A2)}")
# your algorithm time complexity is:O(n)
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:
1. 算法分析：动态规划：定义一个二维数组 $dp[i][w]$，它表示在只考虑前 $i$ 个物品，且背包最大承重为 $w$ 的情况下，所能获得的最大价值。\
&emsp;对于第 $i$ 个物品（重量为 $weight[i-1]$，价值为 $value[i-1]$）：\
&emsp;&emsp; 不装入背包： 当前的最大价值就等于只考虑前 $i-1$ 个物品时的最大价值。$$dp[i][w] = dp[i-1][w]$$
&emsp;&emsp; 装入背包（背包剩余容量足够：$w \ge weight[i-1]$）： 当前的最大价值等于前 $i-1$ 个物品在剩余容量 $w - weight[i-1]$ 下的最大价值，加上当前物品的价值 $value[i-1]$。
 $$dp[i][w] = dp[i-1][w - weight[i-1]] + value[i-1]$$
&emsp;状态转移方程：在两种选择中取最大值：$$dp[i][w] = \max(dp[i-1][w], dp[i-1][w - weight[i-1]] + value[i-1])$$
&emsp;空间优化（一维数组）：因为第 $i$ 行的状态只依赖于第 $i-1$ 行的状态，我们可以将二维数组压缩为一维数组 $dp[w]$。为了防止前面的值被覆盖导致重复放入物品，逆序遍历容量 $w$。\
&emsp;优化后的方程为：$$dp[w] = \max(dp[w], dp[w - weight_i] + value_i)$$

2. 事件复杂度：$O(n \times W)$。\
&emsp;需要遍历 $n$ 个物品，每个物品需要遍历 $W$ 种容量状态，因此时间复杂度是两者的乘积。

````python
# add your code here
# 背包问题
def knapsack(items, capacity):
    # DP数组，长度为capacity+1，初始值全为0
    # dp[w]表示容量为w时能获取的最大价值
    dp = [0] * (capacity + 1)
    
    # 遍历每一个物品
    for weight, value in items:
        # 逆序遍历背包容量，确保每个物品只被放入一次
        # 遍历的下限是当前物品的重量weight，容量小于weight时放不下
        for w in range(capacity, weight - 1, -1):
            # 状态转移
            dp[w] = max(dp[w], dp[w - weight] + value)
            
    # dp 数组的最后一个元素即为最大容量下的最大价值
    return dp[capacity]

# 测试程序
if __name__ == '__main__':
    items = [(5, 10), (4, 40), (6, 30), (3, 50)]
    capacity = 9
    
    max_val = knapsack(items, capacity)
    
    print(f"大理石重量与价值: {items}")
    print(f"小推车最大承重: {capacity}")
    print(f"使用这个小推车装大理石的最大价值为: {max_val}")
# your algorithm time complexity is:O(n*w)
````

## 问题 8

给你二叉搜索树的根节点 root ，该树中的恰好两个节点的值被错误地交换。请在不改变其结构的情况下，恢复这棵树 。设计算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

e.g.:
---
1.

<div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:
1. 算法分析：中序遍历找逆序对
&emsp;对二叉搜索树进行中序遍历得到的节点值序列一定是一个严格递增的序列。如果树中有且仅有两个节点的值被错误交换了，那么破坏了递增性质，导致中序遍历序列中出现逆序对（前一个节点的值大于当前节点的值）。
2. 算法设计步骤：\
&emsp; 维护三个指针：prev：指向上一个遍历到的节点，初始值为极小值、first：第一个被错误交换的节点，初始化为空、second：第二个被错误交换的节点，初始化为空；\
&emsp; 对二叉树进行中序遍历;\
&emsp; 在遍历当前节点 node 时，比较 prev.val 和 node.val：\
&emsp;&emsp;若prev.val>node.val，则找到了逆序对；\
&emsp;&emsp;若first为空，则是第一次遇到逆序对，将prev赋值给first，将node赋值给second；\
&emsp;&emsp;若first不为空，则是第二次遇到逆序对，更新second为当前的node；\
&emsp;遍历结束后，交换first.val和second.val，保持树的结构保持不变。
3. 时间复杂度：$O(N)$\
&emsp;$N$ 是二叉树的节点数。需要进行一次完整的中序遍历来找到被交换的节点，每个节点会被访问一次。

````python
# add your code here
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def __init__(self):
        self.first = None
        self.second = None
        self.prev = TreeNode(float('-inf')) # 初始化前驱节点为一个极小值

    def recoverTree(self, root: TreeNode) -> None:
        def inorder(node: TreeNode):
            if not node:
                return
            
            # 遍历左子树
            inorder(node.left)
            
            # 处理当前节点：寻找逆序对
            if self.prev.val > node.val:
                # 第一次遇到逆序对，记录第一个错误节点prev
                if not self.first:
                    self.first = self.prev
                # 非第一次遇到逆序对，更新第二个错误节点为当前节点node
                self.second = node
            
            # 更新prev指针为当前节点
            self.prev = node
            
            # 遍历右子树
            inorder(node.right)

        # 执行中序遍历
        inorder(root)
        
        # 找到两个错误节点后，交换它们的值
        if self.first and self.second:
            self.first.val, self.second.val = self.second.val, self.first.val

# 将二叉树层次遍历为列表
def treeToList(root):
    if not root: return []
    result = []
    q = [root]
    while q:
        node = q.pop(0)
        if node:
            result.append(node.val)
            q.append(node.left)
            q.append(node.right)
        else:
            result.append(None)
    # 去除末尾多余的None
    while result and result[-1] is None:
        result.pop()
    return result

# 测试程序
if __name__ == '__main__':
    # 构建 root = [1, 3, null, null, 2]
    #      1
    #     /
    #    3
    #     \
    #      2
    root1 = TreeNode(1)
    root1.left = TreeNode(3)
    root1.left.right = TreeNode(2)
    
    print(f"输入1为: root={treeToList(root1)}")
    sol1 = Solution()
    sol1.recoverTree(root1)
    print(f"输出1为: {treeToList(root1)}\n")
    
    # 构建 root = [3, 1, 4, null, null, 2]
    #       3
    #      / \
    #     1   4
    #          \
    #           2
    root2 = TreeNode(3)
    root2.left = TreeNode(1)
    root2.right = TreeNode(4)
    root2.right.left = TreeNode(2)

    print(f"输入2为: root={treeToList(root2)}")
    sol2 = Solution()
    sol2.recoverTree(root2)
    print(f"输出2为: {treeToList(root2)}")
# your algorithm time complexity is:O(N)
````

## 问题 9 \*
给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。



<div style="text-align:center;width:800px;"> <img alt="" src="./fig/hw3q6.png"> </div>

请分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:
1. 算法分析：对于高度数组中的任意位置$i$，该位置能承接的雨水量由其左右两侧最高柱子中的较小值决定，公式为：$$ water[i]=max(0,min(left\_max[i],right\_max[i])−height[i])$$ 其中：
&emsp;&emsp; $left\_max[i]$：位置i左侧（不包含$i$）的最高柱子高度
&emsp;&emsp; $right\_max[i]$：位置i右侧（不包含$i$）的最高柱子高度\
&emsp;&emsp; 总接水量为所有位置接水量的累加和。

2. 算法设计：双指针法：使用两个指针从数组两端向中间移动，同时维护两个变量$left\_max$和$right\_max$：\
&emsp; 如果$height[left] < height[right]$：\
&emsp; &emsp; 若$height[left] >= left\_max$，更新$left\_max$\
&emsp; &emsp; 否则，当前位置接水量为$left\_max - height[left]$，累加到总结果\
&emsp; &emsp; 左指针右移\
&emsp; 否则：\
&emsp; &emsp; 若$height[right] >= right_max$，更新$right\_max$\
&emsp; &emsp; 否则，当前位置接水量为$right\_max - height[right]$，累加到总结果\
&emsp; &emsp; 右指针左移

3. 时间复杂度：\
&emsp;&emsp; 时间复杂度：$O(n)$，仅需一次线性遍历数组

````python
# add your code here
def function(height):
    n = len(height)
    if n < 3:
        return 0
    
    left = 0
    right = n - 1
    left_max = 0
    right_max = 0
    total_water = 0
    
    while left < right:
        if height[left] < height[right]:
            if height[left] >= left_max:
                left_max = height[left]
            else:
                total_water += left_max - height[left]
            left += 1
        else:
            if height[right] >= right_max:
                right_max = height[right]
            else:
                total_water += right_max - height[right]
            right -= 1
    
    return total_water

# 测试程序
# 测试题目示例
if __name__ == "__main__":
    # 输入1
    height1 = [0,1,0,2,1,0,1,3,2,1,2,1]
    print(function(height1))
    
    # 输入2
    height2 = [4,2,0,3,2,5]
    print(function(height2))

# your algorithm time complexity is: O(n)
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:
---
1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:
1. 算法设计思路：深度优先搜索DFS：从根节点一直走到叶子节点，沿途收集节点上的数字。\
&emsp;数字拼接：从上往下走，之前路径代表的数字是 now_sum。当来到当前节点（值为 node.val）时，新的数字就变为：now_sum * 10 + node.val；\
&emsp;终止条件：当到达一个叶子节点时，即一条完整的路径结束了，将当前累积计算出的数字返回；\
&emsp;递归合并：若当前节点不是叶子节点，则继续向左子树和右子树探索，并将左右子树返回的路径和相加，作为当前节点向上的返回值。
2. 时间复杂度: 假设二叉树中总共有 $N$ 个节点，树的高度为 $H$，则时间复杂度：$O(N)$\
&emsp;需要使用深度优先搜索遍历二叉树中的每一个节点。每个节点只会被访问一次，且每次访问时的计算都是常数时间的，因此总时间复杂度为 $O(N)$。

````python
# add your code here
# 二叉树节点类
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def sumNumbers(self, root: TreeNode) -> int:
        def dfs(node, now_sum):  #深度优先遍历
            # 若节点为空，直接返回0
            if not node:
                return 0
            
            # 计算到达当前节点时的数字
            now_sum = now_sum * 10 + node.val
            
            # 若是叶子节点，返回拼接好的数字
            if not node.left and not node.right:
                return now_sum
            
            # 若不是叶子节点，递归计算左右子树的路径和，并相加
            return dfs(node.left, now_sum) + dfs(node.right, now_sum)
        
        # 从根节点开始 DFS，初始数字为0
        return dfs(root, 0)

# 将二叉树层次遍历为列表
def treeToList(root):
    if not root: return []
    result = []
    q = [root]
    while q:
        node = q.pop(0)
        if node:
            result.append(node.val)
            q.append(node.left)
            q.append(node.right)
        else:
            result.append(None)
    # 去除末尾多余的None
    while result and result[-1] is None:
        result.pop()
    return result


# 测试程序
if __name__ == '__main__':
    sol = Solution()
    # 构造用例1
    #      1
    #     / \
    #    2   3
    root1 = TreeNode(1)
    root1.left = TreeNode(2)
    root1.right = TreeNode(3)
    result1 = sol.sumNumbers(root1)
    print(f'输入1: root={treeToList(root1)}', f'输出1: {result1}')

    # 构造用例2
    #      4
    #     / \
    #    9   0
    #   / \
    #  5   1
    root2 = TreeNode(4)
    root2.left = TreeNode(9)
    root2.right = TreeNode(0)
    root2.left.left = TreeNode(5)
    root2.left.right = TreeNode(1)

    result2 = sol.sumNumbers(root2)
    print(f'输入2: root={treeToList(root2)}', f'输出2: {result2}')
# your algorithm time complexity is: O(N)
````
